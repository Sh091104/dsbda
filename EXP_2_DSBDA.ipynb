```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats
```

# Step 2: Create an Academic Performance Dataset

```python
data = {
    'Student_ID': range(1, 21),
    'Name': [f"Student{i}" for i in range(1, 21)],
    'Math_Score': [88, 92, 95, np.nan, 85, 105, 70, 65, 55, 58, 61, 110, 73, 69, 99, 81, np.nan, 67, 79, 95],
    'Science_Score': [75, 82, 89, 91, 60, 65, 88, 70, 72, 69, 68, 66, 78, 74, 105, 102, 58, 54, 93, 98],
    'Attendance_Rate': [0.95, 0.88, 0.90, 0.87, 0.80, 1.02, 0.76, 0.65, np.nan, 0.89, 0.91, 0.85, 0.96, 0.88, 0.92, 0.94, 0.75, 0.86, 0.90, 0.98]
}

df = pd.DataFrame(data)
print(df.head())
```

# Step 3: Handle Missing Values

```python
df['Math_Score'].fillna(df['Math_Score'].mean())
df['Attendance_Rate'].fillna(df['Attendance_Rate'].median())
```

# Step 4: Detect and Handle Outliers

```python
z_scores = np.abs(stats.zscore(df[['Math_Score', 'Science_Score', 'Attendance_Rate']]))
print(np.where(z_scores > 3))
```

```python
# Cap or Remove Outliers (here: capping)
# Capping using IQR
for column in ['Math_Score', 'Science_Score', 'Attendance_Rate']:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    df[column] = np.where(df[column] > upper, upper,
                          np.where(df[column] < lower, lower, df[column]))
```
