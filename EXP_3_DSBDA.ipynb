#  Part 1: Grouped Summary Statistics
# Step 1.1: Import Libraries

```python
import pandas as pd
import numpy as np
```

# Step 1.2: Create a Sample Dataset

```python
data = {
    'age_group': ['18-25', '18-25', '26-35', '26-35', '36-45', '36-45', '46-60', '46-60'],
    'income': [25000, 27000, 35000, 37000, 45000, 43000, 52000, 54000]
}

df = pd.DataFrame(data)
print(df)
```

# Step 1.3: Group by Age Group and Get Summary Statistics

```python
grouped = df.groupby('age_group')['income'].agg(['mean', 'median', 'min', 'max', 'std'])
print(grouped)
```

# Step 1.4: Create a List of Incomes for Each Age Group


```python
income_lists = df.groupby('age_group')['income'].apply(list).to_dict()
print(income_lists)
```
