```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv(r"C:\Users\lenovo\Downloads\Datasets\Datasets\Iris.csv")

# View first few rows
print(df.head())
```

# 1. Feature Types

```python
print("\nFeature Types:\n")
print(df.dtypes)
```

# 2. Histograms of Each Feature

```python
df.drop(columns='Species').hist(figsize=(10, 8), bins=15, edgecolor='black')
plt.suptitle('Histograms of Iris Features', fontsize=16)
plt.tight_layout()
plt.show()
```

# 3. Boxplots of Each Feature

```python
plt.figure(figsize=(10, 6))
df.drop(columns='Species').boxplot()
plt.title('Boxplots of Iris Features')
plt.ylabel('Value (cm)')
plt.xticks(rotation=45)
plt.show()
```
